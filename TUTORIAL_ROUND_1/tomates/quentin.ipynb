```python
import pandas as pd
import plotly.express as px

df1 = pd.read_csv("prices_round_0_day_-1.csv", sep=';')

df1.describe()

df_tomates = df1[df1['product'] == 'TOMATOES'].copy()
df_émmeraudes = df1[df1['product'] == 'EMERALDS'].copy()

fig = px.line(df_tomates, x='timestamp', y='mid_price')
fig.show()

fig2 = px.line(df_émmeraudes, x='timestamp', y='mid_price')
fig2.show()
```

```python
import plotly.graph_objects as go

dt1 = pd.read_csv("trades_round_0_day_-1.csv", sep=';')

dt1.describe()

fig3 = px.line(df_tomates, x='timestamp', y='mid_price')


dtrade_tomate=dt1[dt1['symbol'] == 'TOMATOES'].copy()

fig3.add_trace(
    go.Scatter(
        x=dtrade_tomate['timestamp'],
        y=dtrade_tomate['price'],
        mode = 'markers',
        name='Trade éffectués',
        marker=dict(
            color='red',
            size=3,
            line=dict(color='black',width=1)


                )
    )
)

```

```python
fig4 = px.line(df_tomates, x='timestamp', y=['bid_price_1','ask_price_1'])

fig4.add_trace(
    go.Scatter(
        x=dtrade_tomate['timestamp'],
        y=dtrade_tomate['price'], 
        mode='markers',
        name='Trades éfféctués',
        marker=dict(
            color='green',
            size=3,
            symbol='x',
            line=dict(color='black',width=1)
        )
    )
)


```

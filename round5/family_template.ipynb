# Round 5 — Family Research Template

Set `FAMILY` below, run all. Auto-generates the full report under `round5/reports/<FAMILY>/` and surfaces the key tables inline for quick scrolling. Custom deep-dive cells live at the bottom.

````python
FAMILY = "MICROCHIP"   # one of FAMILIES.keys()
DAYS   = (2, 3, 4)
OUT    = "round5/reports"
````

````python
import sys
from pathlib import Path
ROOT = Path.cwd()
while ROOT.name and not (ROOT / 'round5' / 'research_lib.py').exists():
    if ROOT.parent == ROOT: break
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

from round5 import research_lib as rl
import pandas as pd
pd.set_option('display.max_rows', 60)
pd.set_option('display.max_columns', 30)
pd.set_option('display.width', 220)
````

## Run full family report (writes CSVs + ~20 figures)

````python
out_dir = rl.family_report(FAMILY, out_dir=OUT, days=DAYS)
print('artifacts ->', out_dir)
````

## Inline tables

````python
stats = pd.read_csv(out_dir / 'stats_per_product.csv')
stats.set_index('product').T
````

````python
ic = pd.read_csv(out_dir / 'signals_ic.csv')
pivot = ic.pivot_table(index='signal', columns='product', values='ic_h10')
pivot.style.background_gradient(cmap='RdBu_r', axis=None, vmin=-0.2, vmax=0.2).format('{:+.3f}')
````

````python
corr = pd.read_csv(out_dir / 'corr_mid.csv', index_col=0)
corr.style.background_gradient(cmap='RdBu_r', axis=None, vmin=-1, vmax=1).format('{:+.2f}')
````

````python
leadlag = pd.read_csv(out_dir / 'lead_lag.csv', index_col=0)
leadlag.style.background_gradient(cmap='RdBu_r', axis=None, vmin=-0.2, vmax=0.2).format('{:+.3f}')
````

````python
coint = pd.read_csv(out_dir / 'cointegration.csv')
coint.sort_values('coint_p').head(15)
````
